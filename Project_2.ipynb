# **Visualization**

**Single Line Graph **

**Order_id**

```python
plt.plot(Fast_food_sales['order_id'])
plt.xlabel('Order ID')
plt.ylabel('Transaction Amount')
plt.title('Transaction Amount per Order')
plt.show()
```

**Quantity**

```python
plt.plot(Fast_food_sales['quantity'])
plt.xlabel('Order ID')
plt.ylabel('Quantity Sold')
plt.title('Quantity Sold per Order')
plt.show()
```

**item price**

```python
plt.plot(Fast_food_sales['item_price'])
plt.xlabel('Order ID')
plt.ylabel('Item Price')
plt.title('Item Price per Order')
plt.show()
```

**Multiple Line Graph**

**1.Transaction Amount & Quantity vs Order ID**

```python
plt.plot(Fast_food_sales['order_id'], Fast_food_sales['transaction_amount'], marker='o', label='Transaction Amount')
plt.plot(Fast_food_sales['order_id'], Fast_food_sales['quantity'], marker='s', label='Quantity')
plt.xlabel('Order ID')
plt.title('Transaction Amount and Quantity per Order')
plt.legend()
plt.show()
```

**2.Transaction Amount & Item Price vs Order ID**

```python
plt.plot(Fast_food_sales['order_id'], Fast_food_sales['transaction_amount'], marker='o', label='Transaction Amount')
plt.plot(Fast_food_sales['order_id'], Fast_food_sales['item_price'], marker='^', label='Item Price')
plt.xlabel('Order ID')
plt.title('Transaction Amount and Item Price per Order')
plt.legend()
plt.show()
```

**3.Bar Graph**

**A. Transaction Amount by Item Name**

```python
plt.bar(Fast_food_sales['item_name'], Fast_food_sales['transaction_amount'], color='skyblue')
plt.xlabel('Item Name')
plt.ylabel('Transaction Amount')
plt.title('Transaction Amount by Item Name')
plt.show()
```

**B. Quantity Sold by Item Name**

```python
plt.bar(Fast_food_sales['item_name'], Fast_food_sales['quantity'], color='orange')
plt.xlabel('Item Name')
plt.ylabel('Quantity Sold')
plt.title('Quantity Sold by Item Name')
plt.show()
```

1.**Histrogram**

A. Histogram of Transaction Amount

```python
plt.hist(Fast_food_sales['transaction_amount'], bins=5, color='blue', edgecolor='black')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')
plt.title('Distribution of Transaction Amounts')
plt.show()
```

**B. Histogram of Quantity Sold**

```python
plt.hist(Fast_food_sales['quantity'],bins=5,color='green',edgecolor='black')
plt.xlabel('Quantity Sold')
plt.ylabel('Frequency')
plt.title('Distribution of Transaction Amount')
plt.show()
```

**C. Histogram of Item Price**

```python
plt.hist(Fast_food_sales['item_price'],bins=5,color='red',edgecolor='black')
plt.xlabel('Item Price')
plt.ylabel('Frequency')
plt.title('Distribution of Item Price')
plt.show()
```

**Pie Chart**

**A. Transaction Amount by Item Name**

```python
Amount_by_item = Fast_food_sales.groupby('item_name')['transaction_amount'].sum()

plt.pie(Amount_by_item, labels=Amount_by_item.index, autopct='%1.1f%%', startangle=90)
plt.title('Transaction Amount Distribution by Item Name')
plt.axis('equal')  # ensures perfect circle
plt.show()
```

**B. Quantity Sold by Item Name**

```python
quantity_by_item = Fast_food_sales.groupby('item_name')['quantity'].sum()
plt.pie(quantity_by_item, labels=quantity_by_item.index, autopct='%1.1f%%', startangle=90)
plt.title('Quantity Sold Distribution by Item Name')
plt.axis('equal')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(Fast_food_sales['quantity'], Fast_food_sales['transaction_amount'])
plt.xlabel('Quantity')
plt.ylabel('Transaction Amount')
plt.title('Quantity vs Transaction Amount')
plt.grid(True)
plt.show()
```
